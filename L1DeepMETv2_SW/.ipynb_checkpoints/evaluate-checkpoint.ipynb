```python
import os
import shutil
import sys
import os.path as osp
sys.path
sys.path.append('./L1DeepMETv2/')
from graphmetnetwork import GraphMetNetwork

import numpy as np
import torch

from torch_cluster import radius_graph
from torch_geometric.data import DataLoader
import model.net as net
import model.data_loader as data_loader
import utils
```

### Load Parameters

```python
data_dir = '/usr/scratch/tpham361/'
dataloaders = data_loader.fetch_dataloader(data_dir = data_dir, batch_size=1, validation_split=.2, is_delphes=True)
test_dl = dataloaders['test']
device = torch.device('cpu')
# device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('Test dataloader: {}'.format(len(test_dl)))
print(device)
```

#### Load a sample test data point from the test dataloader

```python
test_data = None
for cnt, test_data in enumerate(test_dl):
    if cnt == 0:
        break
```

```python
test_data.x.shape
```

```python
torch.set_printoptions(sci_mode = False)
print(test_data.x)
```

#### Scale Input Data, Load Tensor Parameters

```python
n_features_cont = 6
x_cont_test = test_data.x[:,:n_features_cont].to(device)  # include puppi
x_cat_test = test_data.x[:,n_features_cont:].long().to(device)
etaphi_test = torch.cat([test_data.x[:, 3][:, None], test_data.x[:, 4][:, None]], dim=1).to(device=device)
batch_test = test_data.batch.to(device)
edge_index_test = radius_graph(etaphi_test, r=0.4, batch=batch_test, loop=False, max_num_neighbors=255).to(device=device)
print(f'x_cont_test: {x_cont_test.shape}')
print(f'x_cat_test: {x_cat_test.shape}')
print(f'etaphi: {etaphi_test.shape}')
print(f'batch: {batch_test.shape}')
print(f'edge_index: {edge_index_test.shape}')
```

#### Convert Tensor parameters to Numpy arrays

```python
x_cont = np.ascontiguousarray(x_cont_test.squeeze(0).cpu().numpy())
x_cat = np.ascontiguousarray(x_cat_test.squeeze(0).cpu().numpy())
batch = np.ascontiguousarray(batch_test.squeeze(0).cpu().numpy())
etaphi = etaphi_test.squeeze(0).cpu().numpy()
edge_index = edge_index_test.squeeze(0).cpu().numpy().transpose()
num_nodes = x_cont.shape[0]
batch_size = batch.shape[0]
print(f'Number of nodes: {num_nodes}')
assert(num_nodes == batch_size)
```

### Load Torch Model and input scaling

```python
prefix = './L1DeepMETv2/ckpts_delphes_20260102'
restore_ckpt = osp.join(prefix, 'last.pth.tar')

scale_momentum = 128
scaling = [1./scale_momentum, 1./scale_momentum, 1./scale_momentum, 1., 1., 1.] # pt, px, py: scale by 128
norm = torch.tensor(scaling).to(device=device)
torch_model = net.Net(continuous_dim=6, categorical_dim=2 , norm=norm, is_delphes=True).to(device)
torch_model.eval()
print(torch_model)
```

#### Get the weights

```python
param_restored_new = utils.load_checkpoint(restore_ckpt, torch_model)
weights_dict = param_restored_new['state_dict']
print(weights_dict)
```

```python
output_dir = "weights_files_Delphes/"

# Check if the directory exists
if os.path.exists(output_dir):
    # Iterate over all the files in the directory
    for filename in os.listdir(output_dir):
        file_path = os.path.join(output_dir, filename)
        try:
            # Check if it's a file and delete it
            if os.path.isfile(file_path) or os.path.islink(file_path):
                os.unlink(file_path)
            # If it's a directory, delete the directory and its contents
            elif os.path.isdir(file_path):
                shutil.rmtree(file_path)
        except Exception as e:
            print(f"Failed to delete {file_path}. Reason: {e}")
else:
    print(f"Directory {output_dir} does not exist.")


# Function to save the weights as binary files
def save_weights_as_binary(weights_dict, output_dir):
    for key, tensor in weights_dict.items():
        # Convert the tensor to a NumPy array
        np_array = tensor.cpu().numpy()

        # Create a binary file name based on the tensor name
        file_name = output_dir + key.replace('.', '_') + '.bin'

        # Save the NumPy array as a binary file
        np_array.tofile(file_name)
        
# Save all weights in the OrderedDict to binary files
save_weights_as_binary(weights_dict, output_dir)
```

### Load C++ Model

```python
output_dir = "weights_files_Delphes/"

# Create an instance of the C++ GraphMetNetwork model
cmodel = GraphMetNetwork()

# Load the weights
cmodel.load_weights(output_dir)
```

#### Test the weights

```python
num_weights = 0
for key, tensor in weights_dict.items():
    # Convert the tensor to a NumPy array
    np_array = tensor.cpu().numpy()

    # Return cmodel function pointer to get the weight array
    cmodel_weight_func_name = 'get_' + key.replace('.', '_')
    cmodel_weight_func = getattr(cmodel, cmodel_weight_func_name)
    cmodel_weight_array = cmodel_weight_func()
    
    # Compare Torch model weight with Cmodel weight
    assert(np.allclose(np_array, cmodel_weight_array, atol=1e-5)), f'cmodel.{cmodel_weight_func_name} returned the wrong weights'
    num_weights += 1

print(f'Number of weights checked: {num_weights}')
```

### Run the Torch Model

```python
num_iters = 110
warmup_iters = 50
test_iters = 20
```

```python
# For GPU

import time
from datetime import datetime

# compiled_torch_model = torch.compile(torch_model, mode = "reduce-overhead")
duration_gpu_all = []

for i in range(warmup_iters):
    with torch.no_grad():
        results = torch_model(x_cont_test, x_cat_test, edge_index_test, batch_test)

for i in range(test_iters):
    start_gpu = torch.cuda.Event(enable_timing = True)
    end_gpu = torch.cuda.Event(enable_timing = True)
    start_gpu.record()
    
    with torch.no_grad():
        results = torch_model(x_cont_test, x_cat_test, edge_index_test, batch_test)
    
    end_gpu.record()
    torch.cuda.synchronize()
    np_results = results.detach().cpu().numpy() # convert result from torch model to numpy array
    duration_gpu = start_gpu.elapsed_time(end_gpu) * 1000 # total duration, in microseconds
    duration_gpu_all.append(duration_gpu)

duration_gpu_all = np.array(duration_gpu_all)
print(f"Device that Torch uses: {torch.cuda.get_device_name()}")
print(f"Model execution duration (GPU): {duration_gpu_all}")
print(f"Model execution duration (GPU): mean = {duration_gpu_all.mean()} us, std = {duration_gpu_all.std()} us")
np.set_printoptions(suppress = True)
print("PyTorch model results:")
print(np_results)
```

```python
import time
from datetime import datetime

# compiled_torch_model = torch.compile(torch_model, mode = "reduce-overhead")

def trace_handler(profiling):
    print(profiling.key_averages().table(sort_by = "self_cuda_time_total"))

with torch.profiler.profile(
    activities = [torch.profiler.ProfilerActivity.CPU, torch.profiler.ProfilerActivity.CUDA],
    schedule = torch.profiler.schedule(wait = 0, warmup = warmup_iters, active = test_iters, repeat = 1),
    on_trace_ready = trace_handler
) as profiling:
    for i in range(num_iters):
        with torch.no_grad():
            results = torch_model(x_cont_test, x_cat_test, edge_index_test, batch_test)
        torch.cuda.synchronize()
        
        np_results = results.detach().cpu().numpy() # convert result from torch model to numpy array
        profiling.step()

print(f"Device that Torch uses: {torch.cuda.get_device_name()}")
np.set_printoptions(suppress = True)
print("PyTorch model results:")
print(np_results)
```

```python
# For CPU

from time import perf_counter_ns
from datetime import datetime

# compiled_cpu_torch_model = torch.compile(torch_model)
duration_cpu_all = []

for i in range(test_iters):
    start_cpu = perf_counter_ns()
    
    with torch.no_grad():
        results = torch_model(x_cont_test, x_cat_test, edge_index_test, batch_test)
    
    end_cpu = perf_counter_ns()
    np_results = results.detach().cpu().numpy() # convert result from torch model to numpy array
    duration_cpu = (end_cpu - start_cpu) * 1000 # total duration, in microseconds
    duration_cpu_all.append(duration_cpu)

duration_cpu_all = np.array(duration_cpu_all)
print(f"Model execution duration (CPU): {duration_cpu_all}")
print(f"Model execution duration (CPU): mean = {duration_cpu_all.mean()} us, std = {duration_cpu_all.std()} us")
np.set_printoptions(suppress = True)
print("PyTorch model results:")
print(np_results)
```

### Run the C++ Model

```python
# import time
import pyRAPL

pyRAPL.setup()
measure = pyRAPL.Measurement("performance")

duration_cpu_all = []
average_power_cpu_all = None

for _ in range(test_iters):
    measure.begin()
    cmodel.GraphMetNetworkLayers(x_cont, x_cat, batch, num_nodes)
    c_results = cmodel.get_output()
    measure.end()

    duration_cpu = measure.result.duration # duration, in us
    energy_cpu_values = measure.result.pkg # list of CPU energy for all sockets, in uJ
    average_power_cpu_values = [value / duration_cpu for value in energy_cpu_values] # list of average power for all sockets, in W = J/s = uJ/us
    
    duration_cpu_all.append(duration_cpu)
    if average_power_cpu_all is None:
        num_sockets = len(average_power_cpu_values)
        average_power_cpu_all = [[] for _ in range(num_sockets)]
    for socket, average_power_cpu in enumerate(average_power_cpu_values):
        average_power_cpu_all[socket].append(average_power_cpu)

duration_cpu_all = np.array(duration_cpu_all)

print(f"Model execution duration (CPU): {duration_cpu_all}")
print(f"Model execution duration (CPU): ({duration_cpu_all.mean()} +/- {duration_cpu_all.std()}) us")
for socket, average_power_cpu_socket in enumerate(average_power_cpu_all):
    average_power_cpu_socket = np.array(average_power_cpu_socket)
    print(f"Average power consumption (CPU - Socket {socket}): ({average_power_cpu_socket.mean()} +/- {average_power_cpu_socket.std()}) W")

print("C model results:")
print(c_results)
# np.testing.assert_allclose(np_results, cmodel.get_output(), rtol=1e-3)
```

### Load all test data points, run both models

```python
import json

datalist_torch = []
datalist_cpp = []

for index, test_data in enumerate(test_dl):
    node_features = test_data.x.tolist()
    print(node_features)
    num_nodes = test_data.x.size(0)

    
    n_features_cont = 6
    x_cont_test = test_data.x[:,:n_features_cont].to(device)  # include puppi
    x_cat_test = test_data.x[:,n_features_cont:].long().to(device)
    etaphi_test = torch.cat([test_data.x[:, 3][:, None], test_data.x[:, 4][:, None]], dim=1).to(device=device)
    batch_test = test_data.batch.to(device)
    edge_index_test = radius_graph(etaphi_test, r=0.4, batch=batch_test, loop=False, max_num_neighbors=255).to(device=device)
    print(f'x_cont_test: {x_cont_test.shape}')
    print(f'x_cat_test: {x_cat_test.shape}')
    print(f'etaphi: {etaphi_test.shape}')
    print(f'batch: {batch_test.shape}')
    print(f'edge_index: {edge_index_test.shape}')

    x_cont = np.ascontiguousarray(x_cont_test.squeeze(0).cpu().numpy())
    x_cat = np.ascontiguousarray(x_cat_test.squeeze(0).cpu().numpy())
    batch = np.ascontiguousarray(batch_test.squeeze(0).cpu().numpy())
    etaphi = etaphi_test.squeeze(0).cpu().numpy()
    edge_index = edge_index_test.squeeze(0).cpu().numpy().transpose()
    num_nodes = x_cont.shape[0]
    batch_size = batch.shape[0]
    print(f'Number of nodes: {num_nodes}')
    assert(num_nodes == batch_size)
    

    model_outputs = torch_model(x_cont_test, x_cat_test, edge_index_test, batch_test)
    results = model_outputs.tolist()
    data_torch = {
        "graph_id": index,
        "num_nodes": num_nodes,
        "node_features": node_features,
        "results": results
    }
    datalist_torch.append(data_torch)

    cmodel.GraphMetNetworkLayers(x_cont, x_cat, batch, num_nodes)
    c_results = cmodel.get_output()
    data_cpp = {
        "graph_id": index,
        "num_nodes": num_nodes,
        "node_features": node_features,
        "results": c_results
    }
    datalist_cpp.append(data_cpp)
    

with open("graph_results_Delphes_torch.json", "w") as file_results_torch:
    json.dump(datalist_torch, file_results_torch, indent = 4)

with open("graph_results_Delphes_cpp.json", "w") as file_results_cpp:
    json.dump(datalist_cpp, file_results_cpp, indent = 4)
```
